Descriptive Statistics & Probability

```python
import numpy as np

usage_hours = np.array([2.4,5.6,7.8,11,4.5,8.9,6.7,9.6,35.0,6.5])

# Mean (Average)
mean_val = usage_hours.mean()  # build in method
mean = np.sum(usage_hours)/len(usage_hours)  # math formula


# Variance
variance = usage_hours.var()      # build in method
var_val = np.sum((usage_hours-mean)**2)/len(usage_hours) # math formula


# Standard Deviation (Data spread)
std_val = usage_hours.std()     # build in method
std = np.sqrt(var_value)       # math formula


print(f"Mean Usage: {mean_val:.2f} hours, =",mean)
print(f"Variance: {variance:.2f}, = ",var_val)
print(f"Standard Deviation: {std_val:.2f}, = ",std)
```

Linear Algebra Foundations: The Dot Product in Neural Networks

```python
input = np.array([2.4,6.7,8.5])

weights = np.array([0.2,0.7,-0.8])

bias = 0.4

# manual dot product + bias { y = w*x +b } core for the linear layer

output = np.sum(input * weights) + bias

# dot product using in build library

dot_prod = np.dot(input,weights) + bias

print(f"Manual Neuron Output: {output:.2f}")
print(f"NumPy Dot Product Output: {dot_prod:.2f}")
```

Matrix Dimensions & Transformations

```python
matrix = np.array([
    [2,4,5],
    [4,1,6],
    [8,7,3]
])  # 3 x 3

vector = np.array([1.5,-0.5,0.9])  # 3 x 1

output_mat = matrix @ vector  # matrix multiplication
print(output_mat)
```

Calculus & Gradient Descent Intuition

```python
# say a convex loss functions as l(w) = (w - 5)^2
# the gradient (derivative) dl/dw = 2 * (w - 5)

w = 0 # initial random guess
learning_rate = 0.2 # step size
epochs = 20

print(f"Starting optimization: Initial w = {w}\n")

for epoch in range(epochs):

    # slope of loss curve
    grad = 2 * (w - 5)

    # new w
    w = w - (learning_rate * grad)

    loss = (w - 5)**2
    print(f"Epoch {epoch+1}: w = {w:.4f}, Loss = {loss:.4f}")

print(f"\nConverged Final w: {w:.4f}")
```
