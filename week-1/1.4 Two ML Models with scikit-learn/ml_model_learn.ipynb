Data loading & preprocessing

```python
import numpy as np 
import pandas as pd

dataset = pd.read_csv("../1.2 data cleaning & EDA/Social_media_impact_on_life.csv")

dataset = dataset.drop_duplicates()
dataset = dataset.dropna()
```

Regression model

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error,r2_score

reg_features = ['Age','Weekend_Extra_Hours','Sleep_Quality_Score','Daily_Usage_Hours','Sleep_Duration_Hours','Perceived_Stress_Score', 'Mental_Health_Index',]

x_reg = dataset[reg_features]
y_reg = dataset["Academic_Performance_GPA"]

# Split
x_train,x_test,y_train,y_test = train_test_split(x_reg,y_reg,test_size=0.2,random_state=42)

regression_pipeline = Pipeline([
    ('Scaler',StandardScaler()),
    ('model',LinearRegression())
])

# train 
regression_pipeline.fit(x_train,y_train)

# prediction
y_predict = regression_pipeline.predict(x_test)

# Evaluate

mse = mean_squared_error(y_test,y_predict)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_predict)

print("--- Regression Model Evaluation (GPA Prediction) ---")
print(f"RMSE: {rmse:.4f}")
print(f"R-squared (R2): {r2:.4f}")

```

Classification Model

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
from sklearn.ensemble import RandomForestClassifier

features = ['Age','Academic_Performance_GPA','Weekend_Extra_Hours','Sleep_Quality_Score','Daily_Usage_Hours','Sleep_Duration_Hours','Perceived_Stress_Score', 'Mental_Health_Index',]

x_features = dataset[features]
y_labels = dataset["Overall_Impact"]


# split
X_train,X_test,Y_train,Y_test = train_test_split(x_features,y_labels,test_size=0.2,random_state=42)


classification_pL= Pipeline([
    ('Scaler',StandardScaler()),
    ('model',RandomForestClassifier(n_estimators=100,random_state=42))
])

# train
classification_pL.fit(X_train,Y_train)

# predict
Y_pred = classification_pL.predict(X_test)

# Evaluate
acc = accuracy_score(Y_test,Y_pred)
report = classification_report(Y_test,Y_pred)

print("--- Classification Model Evaluation ---")
print(f"Accuracy Score: {acc:.2f}")
print(f"Classification report: {report}")

```
