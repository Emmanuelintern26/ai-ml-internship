```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
import torch.optim as optim
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader,Dataset
```

```python
data = pd.read_csv("Social_media_impact_on_life.csv")
data.drop_duplicates(inplace=True)
data.dropna(inplace=True)


features = ['Age','Weekend_Extra_Hours','Sleep_Quality_Score','Daily_Usage_Hours','Sleep_Duration_Hours','Perceived_Stress_Score', 'Mental_Health_Index',]

x = data[features].values
y = data["Academic_Performance_GPA"].values.reshape(-1,1) # into column vector


# scale features
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

# split
X_train,X_test,y_train,y_test = train_test_split(x_scaled,y,test_size=0.2,random_state=42)

```

```python
# custom pytorch dataset

class StudentDataset(Dataset):

    def __init__(self, features, targets):
        self.features = torch.tensor(features, dtype=torch.float32)
        self.targets = torch.tensor(targets, dtype=torch.float32)

    def __len__(self):
        return len(self.features)

    def __getitem__(self, idx):
        return self.features[idx], self.targets[idx]

train_data = StudentDataset(X_train,y_train)
test_data = StudentDataset(X_test,y_test)

#Data loading 
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64, shuffle=False)

```
